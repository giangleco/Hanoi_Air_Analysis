Sử dụng Python và các thư viện phân tích dữ liệu

## 2. Import thư viện và đọc dữ liệu

Ở phần này, sử dụng các thư viện Python để xử lý,
phân tích và trực quan hóa bộ dữ liệu.

```python
# Import các thư viện
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Cấu hình hiển thị biểu đồ
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (12, 6)

# Đọc dữ liệu
df = pd.read_csv("Data/hanoi-aqi-weather-data.csv")

# Chuyển Local Time sang kiểu datetime
df["Local Time"] = pd.to_datetime(df["Local Time"])

# Sắp xếp dữ liệu theo thời gian
df = df.sort_values("Local Time")

# Đặt Local Time làm chỉ mục
df = df.set_index("Local Time")

# Hiển thị 5 dòng đầu tiên
display(df.head())
```

## 3. Khám phá và kiểm tra dữ liệu

Trước khi phân tích, cần kiểm tra cấu trúc,
kiểu dữ liệu và chất lượng của bộ dữ liệu.

```python
# Kích thước dữ liệu
print("Số dòng:", df.shape[0])
print("Số cột:", df.shape[1])

# Kiểu dữ liệu
display(df.dtypes)

# Thông tin tổng quan
df.info()

# Kiểm tra giá trị thiếu
print("Số giá trị thiếu:")
display(df.isnull().sum())

# Kiểm tra dữ liệu trùng lặp
print("Số dòng trùng lặp:", df.duplicated().sum())

# Kiểm tra thời gian
print("Thời gian bắt đầu:", df.index.min())
print("Thời gian kết thúc:", df.index.max())
```

### Nhận xét

Bộ dữ liệu có 8.785 bản ghi và 19 thuộc tính.
Dữ liệu bao gồm thông tin thời gian, địa điểm,
các chỉ số chất lượng không khí và yếu tố thời tiết.

Cần tiếp tục kiểm tra dữ liệu trước khi thực hiện
các phân tích thống kê và trực quan hóa.

## 4. Thống kê mô tả

Thống kê mô tả giúp khái quát đặc điểm của dữ liệu
thông qua các giá trị trung bình, độ lệch chuẩn,
giá trị nhỏ nhất, lớn nhất và các phân vị.

```python
# Lựa chọn các cột dữ liệu số
numeric_df = df.select_dtypes(include=np.number)

# Thống kê mô tả
display(numeric_df.describe())
```

```python
print("AQI trung bình:", df["AQI"].mean())
print("PM2.5 trung bình:", df["PM25"].mean())
print("PM10 trung bình:", df["PM10"].mean())
print("Nhiệt độ trung bình:", df["Temperature"].mean())
print("Độ ẩm trung bình:", df["Relative Humidity"].mean())
```

## 5. Phân tích chỉ số chất lượng không khí AQI

AQI (Air Quality Index) là chỉ số thể hiện
mức độ chất lượng không khí.

Phần này tập trung quan sát sự biến động của AQI
trong toàn bộ khoảng thời gian nghiên cứu.

```python
plt.figure(figsize=(15, 6))

plt.plot(df.index, df["AQI"], linewidth=0.7)

plt.title("Biến động chỉ số AQI tại Hà Nội")
plt.xlabel("Thời gian")
plt.ylabel("Chỉ số AQI")

plt.axhline(50, linestyle="--", label="AQI = 50")
plt.axhline(100, linestyle="--", label="AQI = 100")
plt.axhline(150, linestyle="--", label="AQI = 150")
plt.axhline(300, linestyle="--", label="AQI = 300")

plt.legend()
plt.tight_layout()
plt.show()
```

```python
# 10 thời điểm có AQI cao nhất
top_aqi = df["AQI"].nlargest(10)

display(top_aqi)
```

```python
plt.figure(figsize=(15, 6))

plt.plot(df.index, df["PM25"], linewidth=0.7)

plt.title("Biến động nồng độ PM2.5 tại Hà Nội")
plt.xlabel("Thời gian")
plt.ylabel("PM2.5 (µg/m³)")

plt.tight_layout()
plt.show()
```

```python
# Tính PM2.5 trung bình theo từng giờ
hourly_pm25 = df.groupby(df.index.hour)["PM25"].mean()

plt.figure(figsize=(12, 6))

sns.barplot(
    x=hourly_pm25.index,
    y=hourly_pm25.values
)

plt.title("PM2.5 trung bình theo giờ trong ngày")
plt.xlabel("Giờ")
plt.ylabel("PM2.5 trung bình (µg/m³)")

plt.xticks(range(24))
plt.tight_layout()
plt.show()
```

```python
# Tạo cột tháng
df["Month"] = df.index.month

# Phân loại mùa
def classify_season(month):
    if month in [2, 3, 4]:
        return "Xuân"
    elif month in [5, 6, 7]:
        return "Hạ"
    elif month in [8, 9, 10]:
        return "Thu"
    else:
        return "Đông"

df["Season"] = df["Month"].apply(classify_season)

# Tính PM2.5 trung bình theo mùa
season_order = ["Xuân", "Hạ", "Thu", "Đông"]

seasonal_pm25 = (
    df.groupby("Season")["PM25"]
    .mean()
    .reindex(season_order)
)

display(seasonal_pm25)
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Temperature",
    y="PM25",
    alpha=0.4
)

plt.title("Mối quan hệ giữa nhiệt độ và PM2.5")
plt.xlabel("Nhiệt độ (°C)")
plt.ylabel("PM2.5 (µg/m³)")

plt.tight_layout()
plt.show()
```

```python
weather_columns = [
    "Precipitation",
    "Pressure",
    "Relative Humidity",
    "Wind Speed"
]

for column in weather_columns:

    plt.figure(figsize=(9, 5))

    sns.scatterplot(
        data=df,
        x=column,
        y="PM25",
        alpha=0.4
    )

    plt.title(f"Mối quan hệ giữa {column} và PM2.5")
    plt.xlabel(column)
    plt.ylabel("PM2.5 (µg/m³)")

    plt.tight_layout()
    plt.show()
```

## 8. Phân tích tương quan

Hệ số tương quan Pearson được sử dụng để đo lường
mức độ và chiều hướng của mối quan hệ tuyến tính
giữa hai biến số.

```python
correlation_columns = [
    "PM25",
    "Precipitation",
    "Pressure",
    "Relative Humidity",
    "Temperature",
    "Wind Speed"
]

correlation_matrix = df[correlation_columns].corr()

display(correlation_matrix)
```

```python
plt.figure(figsize=(10, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Ma trận tương quan giữa PM2.5 và các yếu tố thời tiết")

plt.tight_layout()
plt.show()
```
